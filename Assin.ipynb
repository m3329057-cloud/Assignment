```python
#Quadratic collion
#Given array
arr=[23,43,13,27,37]
#Hash table size
size=10
#Define hash talbe
hash_table=[None]*size
#Insert every umber
for num in arr:
    #Calculate hash index
    index=num%size
    i=1
    #if index is already accoupied
    while hash_table[index] is not None:
        #Move to the next index
        index=(num%size + i+i)%size
        i+=1
        break
    #Store the number at empty index
    hash_table[index]=num
#Print the hash talbe
print(hash_table)
```
